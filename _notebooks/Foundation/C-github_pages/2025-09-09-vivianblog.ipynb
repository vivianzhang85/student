# Blog entry 2

## 😞 Debugging headaches: Make, permalinks, and my disappearing snake

### I hit two snags this week while updating my snake game blog.

#### 1. Make not cooperating
#### I kept trying to rebuild with make, but nothing was updating. Turns out I had misconfigured my build rules — basically, my changes weren’t even getting processed. Once I cleaned up my Makefile and re-ran properly, things finally built.

#### 2. Permalink issue (forgot the date)
#### I also learned that Jekyll needs a date in the frontmatter (YYYY-MM-DD-title.md) or the permalink breaks. Without it, my /snake/ page wasn’t resolving. Adding the date fixed it instantly.

#### 🐍 3. Snake color hack deleting the snake
#### When I tried to change my snake’s color, the whole snake vanished. The problem? My color-change code was tied to game initialization, which reset the snake array. The fix was simple: update snakeColor inside the draw function and avoid re-calling initGame(). Now I can swap colors live without nuking the snake. 

#### Takeaway: 
#### Don’t overcomplicate — keep build configs clean, frontmatter consistent, and separate UI tweaks (like color) from state resets (like starting a new game).

## Leveling up my Snake game
#### I’ve been experimenting with my Snake game and added a few fun twists to make it feel less like the basic classroom version and more like a mini-arcade project.

## ✨ New features I built
#### Glow effect: the snake now has a neon-like glow when it moves, which makes the game more visually fun (and a little chaotic).Speed boost: grabbing certain food makes the snake move faster — so you get an adrenaline spike along with your score.Triple growth: instead of only growing by 1 block, I added a mode where the snake grows by 3 segments at a time. It ramps up difficulty fast and forces new strategies.

## 🐍 What I learned
#### Separating UI changes (like color and glow) from game state (the snake’s body array) is key. If you reset both together, the snake disappears. Small tweaks in mechanics (like growth size or speed) can totally change how the game feels to play.

## 💡 Takeaway
#### Even simple games become way more engaging with just a few feature experiments. Now my Snake isn’t just crawling along — it glows, dashes, and bulks up.

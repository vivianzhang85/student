# Blog Entry 2

## 😞 Debugging Headaches: Make, Permalinks, and My Disappearing Snake

This week, I ran into several frustrating issues while updating my Snake game blog.  
Debugging felt like a full-time mini-project, but I ended up learning a lot about build tools, Jekyll, and game state management.

### 1. Make Not Cooperating
I kept trying to rebuild my project using `make`, but nothing seemed to update.  

**The problem:** My build rules were misconfigured, so changes weren’t being processed at all.  

**The fix:** After carefully reviewing and cleaning up my `Makefile`, I re-ran `make`, and everything built correctly.  

**Lesson learned:** Always double-check build scripts when updates aren’t reflecting. Sometimes the issue isn’t the code — it’s the tools around it.

### 2. Permalink Issue (Forgot the Date)
My `/snake/` page wasn’t resolving correctly on Jekyll.  

**The problem:** Jekyll requires a date in the frontmatter (`YYYY-MM-DD-title.md`) for permalinks to work. I had forgotten to include it.  

**The fix:** Adding the date to the filename immediately fixed the issue.  

**Lesson learned:** Small details in static site generators matter a lot — missing frontmatter can break pages in confusing ways.

### 3. 🐍 Snake Color Hack Deleting the Snake
I tried to add a fun feature: changing the snake’s color dynamically. Instead, the snake disappeared completely.  

**The problem:** My color-change code was tied to game initialization, which reset the snake array every time I applied a new color.  

**The fix:** I moved the `snakeColor` update inside the draw function and avoided re-calling `initGame()`.  
Now I can swap colors live without wiping the snake’s progress.  

**Lesson learned:** Separating UI tweaks from game state is essential. Otherwise, minor changes can unintentionally break core mechanics.

### 💡 Takeaway
Debugging taught me to:  
- Keep build configurations clean and simple.  
- Ensure frontmatter is consistent in Jekyll projects.  
- Separate visual or UI tweaks from core game logic.  

Even small mistakes in tooling or state management can cause disproportionately large headaches — but each fix is a learning opportunity that makes the project stronger.


## Blog Entry 3
## Leveling Up My Snake Game
I’ve been experimenting with my Snake game and added a few fun twists to make it feel less like the basic classroom version and more like a mini-arcade project.  
Over the past week, I focused on adding visual flair, gameplay mechanics, and small experiments that make the game more dynamic and exciting.  

## ✨ New Features I Built

### Glow Effect
The snake now has a neon-like glow when it moves.  
This simple visual tweak makes the game more fun to watch and play, giving it a lively, arcade-style feel. The glow also helps track the snake’s movement, especially as it grows longer.

### Speed Boost
I added certain "power foods" that temporarily increase the snake’s speed when eaten.  
This gives players an adrenaline spike along with their score, making the gameplay more thrilling and requiring faster reflexes. Strategically timing these boosts adds a layer of skill to the game.

### Triple Growth
Instead of only growing by 1 block per food, the snake can now grow by 3 segments at once in a special mode.  
This ramps up difficulty quickly, forcing players to adapt their strategy and plan ahead. It’s amazing how just a small tweak in growth mechanics can completely change the pacing of the game.

## 🐍 What I Learned
Working on these features reinforced a few important programming lessons:

- **Separate UI from game state:** Changes to visuals, like color or glow, should not interfere with the snake’s core logic (its body array). Resetting both together can make the snake disappear entirely.  
- **Small changes can have big effects:** Adjusting speed, growth rate, or power-up mechanics drastically alters gameplay, demonstrating the importance of careful experimentation.  
- **Iterate quickly:** Testing features in small increments helps catch unexpected bugs and ensures each change improves the player experience.

## 💡 Takeaway
Even simple games become way more engaging with a few thoughtful feature experiments.  
By combining visual effects, gameplay mechanics, and strategic tweaks, my Snake game has evolved from a basic classroom project into a mini-arcade experience.  

Now, my Snake isn’t just crawling along — it **glows**, **dashes**, and **bulks up**, giving players a more exciting and challenging experience with every playthrough.

```python
%%js
<!DOCTYPE html>
<html lang="en">
<head>
<meta charset="UTF-8">
<title>JS Random Hacks - Coin & Compliment</title>
<style>
/* 🕹️ JS Random Homework */
.js-random-hw {
  background: linear-gradient(135deg, #0d001a, #1a0033, #220044);
  border: 2px solid #ff4081;
  border-radius: 14px;
  padding: 30px;
  margin: 30px auto;
  max-width: 600px;
  font-family: "Segoe UI", sans-serif;
  color: #f9eaff;
  box-shadow: 0 0 35px rgba(255, 64, 129, 0.7);
  line-height: 1.8em;
}

.js-random-hw h2 {
  color: #ff4081;
  text-shadow: 0 0 15px rgba(255, 64, 129, 1);
  font-family: "Orbitron", "Segoe UI", sans-serif;
  margin-bottom: 15px;
  font-size: 1.8em;
}

.js-random-hw h3 {
  color: #00e5ff;
  text-shadow: 0 0 10px rgba(0, 229, 255, 0.9);
  margin-top: 18px;
}

button {
  margin-top: 8px;
  padding: 8px 14px;
  border: none;
  border-radius: 8px;
  font-weight: bold;
  cursor: pointer;
  background: #ff4081;
  color: white;
  transition: all 0.2s ease-in-out;
}
button:hover { background: #00e5ff; color: #000; }

.output {
  margin: 8px;
  font-size: 1.2em;
  color: #ffeb3b;
}
</style>
</head>
<body>

<div class="js-random-hw">
  <h2>JS Random Hacks - Coin & Compliment</h2>

  <!-- Hack 3: Coin Flip Streak Tracker -->
  <h3>Coin Flip Streak Tracker</h3>
  <button onclick="flipCoin()">Flip Coin</button>
  <p id="coinResult" class="output"></p>
  <p id="coinStreak" class="output"></p>

  <!-- Hack 4: Random Compliment Generator -->
  <h3>Random Compliment Generator</h3>
  <button onclick="newCompliment()">Get Compliment</button>
  <p id="complimentOut" class="output"></p>
</div>

<script>
// --- Hack 3: Coin Flip Streak Tracker ---
let lastFlip = "";
let currentStreak = 0;
let longestStreak = 0;

function flipCoin() {
  const flip = Math.random() < 0.5 ? "Heads" : "Tails";
  if (flip === lastFlip) currentStreak++;
  else currentStreak = 1;
  lastFlip = flip;
  if (currentStreak > longestStreak) longestStreak = currentStreak;
  document.getElementById("coinResult").textContent = flip;
  document.getElementById("coinStreak").textContent = `Current Streak: ${currentStreak} | Longest Streak: ${longestStreak}`;
}

// --- Hack 4: Random Compliment Generator ---
const compliments = [
  "You are a coding superstar.",
  "Your ideas are sharp and creative.",
  "You make hard things look easy.",
  "You have great problem-solving skills.",
  "Your attention to detail is impressive.",
  "You are always improving.",
  "You are a strong thinker.",
  "Your persistence pays off.",
  "You learn quickly.",
  "You bring good energy to projects."
];

function newCompliment() {
  const rand = Math.floor(Math.random() * compliments.length);
  document.getElementById("complimentOut").textContent = compliments[rand];
}
</script>

</body>
</html>
```

```python
%%js
<!DOCTYPE html>
<html lang="en">
<head>
<meta charset="UTF-8" />
<meta name="viewport" content="width=device-width, initial-scale=1" />
<title>Hack 3 & 4</title>
<style>
  body {
    font-family: "Segoe UI", Tahoma, Geneva, Verdana, sans-serif;
    background: #120022;
    color: #f9eaff;
    display: flex;
    flex-direction: column;
    align-items: center;
    padding: 30px;
    min-height: 100vh;
  }
  .box {
    background: linear-gradient(135deg, #0d001a, #1a0033, #220044);
    border: 2px solid #ff4081;
    border-radius: 14px;
    padding: 30px;
    margin: 20px;
    max-width: 400px;
    width: 100%;
    box-shadow: 0 0 35px rgba(255, 64, 129, 0.7);
  }
  h2 {
    color: #ff4081;
    text-shadow: 0 0 15px rgba(255, 64, 129, 1);
  }
  button {
    background: #ff4081;
    border: none;
    color: white;
    padding: 12px 25px;
    border-radius: 10px;
    font-weight: bold;
    cursor: pointer;
    box-shadow: 0 0 10px #ff4081;
    font-size: 1.1rem;
  }
  button:hover {
    background: #e0316f;
  }
  .result {
    margin-top: 15px;
    font-size: 1.2rem;
    min-height: 3em;
  }
  .compliment {
    color: #76ff03;
    text-shadow: 0 0 8px rgba(118, 255, 3, 0.9);
  }
  .streak {
    color: #ff9800;
    text-shadow: 0 0 8px rgba(255, 152, 0, 1);
  }
</style>
</head>
<body>

<div class="box" id="hack3">
  <h2>Hack 3: Coin Flip Streak Tracker</h2>
  <button id="flipBtn">Flip Coin</button>
  <div class="result streak" id="flipResult">Click flip to start!</div>
  <div class="result streak" id="currentStreak">Current streak: 0</div>
  <div class="result streak" id="longestStreak">Longest streak: 0</div>
</div>

<div class="box" id="hack4">
  <h2>Hack 4: Random Compliment Generator</h2>
  <button id="complimentBtn">Get Compliment</button>
  <div class="result compliment" id="complimentText">You’re awesome!</div>
</div>

<script>
  // Hack 3: Coin Flip Streak Tracker
  const flipBtn = document.getElementById("flipBtn");
  const flipResult = document.getElementById("flipResult");
  const currentStreakEl = document.getElementById("currentStreak");
  const longestStreakEl = document.getElementById("longestStreak");

  let lastFlip = null;
  let currentStreak = 0;
  let longestStreak = 0;

  flipBtn.addEventListener("click", () => {
    const flip = Math.random() < 0.5 ? "Heads" : "Tails";

    flipResult.textContent = `You flipped: ${flip}`;

    if (flip === lastFlip) {
      currentStreak++;
    } else {
      currentStreak = 1;
      lastFlip = flip;
    }

    if (currentStreak > longestStreak) {
      longestStreak = currentStreak;
    }

    currentStreakEl.textContent = `Current streak: ${currentStreak}`;
    longestStreakEl.textContent = `Longest streak: ${longestStreak}`;
  });


  // Hack 4: Random Compliment Generator
  const complimentBtn = document.getElementById("complimentBtn");
  const complimentText = document.getElementById("complimentText");

  const compliments = [
    "You’re an incredible person!",
    "Your smile lights up the room.",
    "You have a great sense of humor!",
    "You make a difference every day.",
    "Your creativity is inspiring.",
    "You’re so thoughtful and kind.",
    "You have amazing ideas!",
    "You brighten everyone’s day.",
    "Your positivity is contagious.",
    "You’re doing great — keep it up!",
    "You’re a fantastic friend.",
    "You have a brilliant mind!",
    "Your energy is inspiring.",
    "You’re a true gem.",
    "You’re simply amazing!"
  ];

  complimentBtn.addEventListener("click", () => {
    const randomIndex = Math.floor(Math.random() * compliments.length);
    complimentText.textContent = compliments[randomIndex];
  });
</script>

</body>
</html>
```

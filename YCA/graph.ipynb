```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from scipy import stats
import os

# Read the CSV file
file_path = os.path.expanduser('~/YCA/cpFeb 2026 indoor vs outdoor.xls - all data.csv')
df = pd.read_csv(file_path)

# Convert time column to datetime if it's not already
# You may need to adjust the column name for time based on your actual data
time_column = df.columns[0]  # Assuming time is the first column, adjust as needed
df[time_column] = pd.to_datetime(df[time_column])

# Sort by time to ensure proper plotting
df = df.sort_values(time_column)

# Create the scatter plot
fig, axes = plt.subplots(2, 1, figsize=(12, 10))

# Plot 1: PM2.5 vs Time
axes[0].scatter(df[time_column], df['pm2.5'], alpha=0.5, label='PM2.5 Data', color='blue')

# Add line of best fit for PM2.5
# Convert time to numeric for regression
time_numeric = (df[time_column] - df[time_column].min()).dt.total_seconds()
z = np.polyfit(time_numeric, df['pm2.5'], 1)
p = np.poly1d(z)
axes[0].plot(df[time_column], p(time_numeric), "r-", linewidth=2, label=f'Best Fit Line (slope: {z[0]:.4f})')

axes[0].set_xlabel('Time')
axes[0].set_ylabel('PM2.5 Concentration')
axes[0].set_title('PM2.5 vs Time with Line of Best Fit')
axes[0].legend()
axes[0].grid(True, alpha=0.3)
axes[0].tick_params(axis='x', rotation=45)

# Plot 2: PM1.0 vs Time
axes[1].scatter(df[time_column], df['pm1.0'], alpha=0.5, label='PM1.0 Data', color='green')

# Add line of best fit for PM1.0
z_pm1 = np.polyfit(time_numeric, df['pm1.0'], 1)
p_pm1 = np.poly1d(z_pm1)
axes[1].plot(df[time_column], p_pm1(time_numeric), "r-", linewidth=2, label=f'Best Fit Line (slope: {z_pm1[0]:.4f})')

axes[1].set_xlabel('Time')
axes[1].set_ylabel('PM1.0 Concentration')
axes[1].set_title('PM1.0 vs Time with Line of Best Fit')
axes[1].legend()
axes[1].grid(True, alpha=0.3)
axes[1].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()

# Optional: Print statistics
print("\n=== Statistics ===")
print(f"PM2.5 - Correlation with time: {stats.pearsonr(time_numeric, df['pm2.5'])[0]:.4f}")
print(f"PM2.5 - P-value: {stats.pearsonr(time_numeric, df['pm2.5'])[1]:.4f}")
print(f"PM2.5 - Slope: {z[0]:.4f}")
print(f"PM2.5 - Intercept: {z[1]:.4f}")
print(f"\nPM1.0 - Correlation with time: {stats.pearsonr(time_numeric, df['pm1.0'])[0]:.4f}")
print(f"PM1.0 - P-value: {stats.pearsonr(time_numeric, df['pm1.0'])[1]:.4f}")
print(f"PM1.0 - Slope: {z_pm1[0]:.4f}")
print(f"PM1.0 - Intercept: {z_pm1[1]:.4f}")
```
